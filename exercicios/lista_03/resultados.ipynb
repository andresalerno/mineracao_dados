## Exercício 1 — Distância em passos de desvio-padrão

- Calcule a distância entre o valor e a média.
- Descubra quantos desvios-padrão cabem nessa distância.
- Mostre o Z-Score.
- Exiba uma frase interpretando o resultado em linguagem natural.

```python
media = 100
desvio = 5
valor = 115

distancia = valor - media
n_desvios = distancia / desvio
z = (valor - media) / desvio

print(f"Distância entre o valor e a média: {distancia}")
print(f"Quantidade de desvios-padrão contidos na distância: {n_desvios}")
print(f"Z-Score: {z}")
print(f"Interpretação: o valor {valor} está {abs(z)} desvios-padrão acima da média ({media}).")
```

## Exercício 2 — Acima ou abaixo da média?

- Calcule o Z-Score dos três valores.
- Para cada resultado, informe se o valor está abaixo da média, exatamente na média ou acima da média.
- Explique o que significa o sinal positivo, negativo ou zero.

```python
casos = [85, 100, 120]
media = 100
desvio = 10

for valor in casos:
    z = (valor - media) / desvio
    if z < 0:
        posicao = "abaixo da média"
    elif z > 0:
        posicao = "acima da média"
    else:
        posicao = "exatamente na média"
    print(f"Valor {valor}: Z = {z:.2f} -> está {posicao}")
```

**Interpretação:** o sinal do Z-Score indica a direção em relação à média — negativo significa
que o valor está abaixo da média, positivo significa que está acima, e zero significa que o valor
coincide exatamente com a média. A magnitude (valor absoluto) indica o quão longe da média o valor está,
independente da direção.

## Exercício 3 — Qual leitura é mais incomum?

- Calcule o Z-Score de cada temperatura.
- Identifique qual valor está mais distante da média em termos de desvio-padrão.
- Mostre qual possui o maior valor absoluto de Z.

```python
temperaturas = [21, 23, 25, 27, 29]
media = 25
desvio = 2

zs = [(t - media) / desvio for t in temperaturas]

for t, z in zip(temperaturas, zs):
    print(f"Temperatura {t}°C -> Z = {z:.2f}")

idx_mais_incomum = max(range(len(zs)), key=lambda i: abs(zs[i]))
print()
print(f"Leitura mais incomum: {temperaturas[idx_mais_incomum]}°C, "
      f"com Z = {zs[idx_mais_incomum]:.2f} (|Z| = {abs(zs[idx_mais_incomum]):.2f})")
```

## Exercício 4 — Latência de uma API

- Calcule a média e o desvio-padrão do conjunto.
- Calcule somente o Z-Score da latência 180 ms.
- Informe se esse valor merece investigação usando |Z| > 3.
- Mostre uma mensagem lembrando que investigar não significa apagar automaticamente.

```python
import numpy as np

latencias = [98, 102, 101, 99, 100, 103, 97, 180]

media = np.mean(latencias)
desvio = np.std(latencias)

z_180 = (180 - media) / desvio

print(f"Média: {media:.2f} ms")
print(f"Desvio-padrão: {desvio:.2f} ms")
print(f"Z-Score de 180 ms: {z_180:.2f}")

if abs(z_180) > 3:
    print("A latência de 180 ms merece investigação (|Z| > 3).")
else:
    print("A latência de 180 ms não ultrapassa o limite de |Z| > 3.")

print("Lembrete: investigar não significa apagar o dado automaticamente — "
      "o valor pode ter uma causa real que precisa ser entendida.")
```

## Exercício 5 — Monitoramento de CPU com classificação

- Calcule a média e o desvio-padrão.
- Percorra todas as leituras.
- Calcule o Z-Score de cada uma.
- Classifique cada leitura como Comum ou Investigar.
- Mostre o resultado no formato valor -> Z-Score -> classificação.

```python
import numpy as np

cpu = [42, 45, 47, 44, 46, 43, 48, 92]

media = np.mean(cpu)
desvio = np.std(cpu)

print(f"Média: {media:.2f}")
print(f"Desvio-padrão: {desvio:.2f}")
print()

for valor in cpu:
    z = (valor - media) / desvio
    classificacao = "Investigar" if abs(z) > 3 else "Comum"
    print(f"{valor} -> {z:.2f} -> {classificacao}")
```

## Exercício 6 — O mesmo valor em dois contextos

- Calcule o Z-Score de 110 no Grupo A.
- Calcule o Z-Score de 110 no Grupo B.
- Compare os resultados.
- Explique por que a mesma distância absoluta de 10 unidades pode ser muito incomum em um grupo e comum em outro.

```python
media_a = 100
desvio_a = 2

media_b = 100
desvio_b = 20

valor = 110

z_a = (valor - media_a) / desvio_a
z_b = (valor - media_b) / desvio_b

print(f"Grupo A -> Z = {z_a:.2f}")
print(f"Grupo B -> Z = {z_b:.2f}")
```

**Interpretação:** o valor 110 está 10 unidades acima da média nos dois grupos, mas o Z-Score
mostra o quão incomum essa distância realmente é. No Grupo A, o desvio-padrão é pequeno (2), então
10 unidades representam 5 desvios-padrão — um valor extremamente raro. No Grupo B, o desvio-padrão é
grande (20), então a mesma distância de 10 unidades representa apenas 0.5 desvio-padrão — algo bem
comum. Isso mostra que a distância absoluta, sozinha, não diz nada sobre o quão incomum um valor é;
é preciso relativizar essa distância pela variabilidade natural do próprio grupo.

## Exercício 7 — Função de interpretação

- Complete a função para receber um Z-Score.
- Se |Z| > 3, retorne "Investigar".
- Se Z < 0, retorne "Abaixo da média".
- Se Z > 0, retorne "Acima da média".
- Se Z == 0, retorne "Na média".
- Teste a função com todos os valores fornecidos.

```python
def interpretar_z(z):
    if abs(z) > 3:
        return "Investigar"
    elif z < 0:
        return "Abaixo da média"
    elif z > 0:
        return "Acima da média"
    else:
        return "Na média"

valores_z = [-3.5, -1.2, 0, 0.8, 3.7]

for z in valores_z:
    print(f"Z = {z} -> {interpretar_z(z)}")
```

## Exercício 8 — Z-Score em um DataFrame

- Calcule a média da coluna Requisicoes.
- Calcule o desvio-padrão da coluna.
- Crie a coluna Z_Score.
- Crie a coluna Status com os valores Comum ou Investigar.
- Exiba apenas as linhas marcadas para investigação.

```python
import pandas as pd
import numpy as np

dados = {
    "Usuario": ["ana", "bruno", "carla", "diego", "eva", "fabio"],
    "Requisicoes": [120, 135, 128, 122, 130, 400]
}

df = pd.DataFrame(dados)

media = df["Requisicoes"].mean()
desvio = df["Requisicoes"].std(ddof=0)

df["Z_Score"] = (df["Requisicoes"] - media) / desvio
df["Status"] = df["Z_Score"].apply(lambda z: "Investigar" if abs(z) > 3 else "Comum")

print(f"Média: {media:.2f}")
print(f"Desvio-padrão: {desvio:.2f}")
print()
print("DataFrame completo:")
print(df)
print()
print("Linhas marcadas para investigação:")
print(df[df["Status"] == "Investigar"])
```

## Exercício 9 — Comparação entre IQR e Z-Score

- Calcule Q1, Q3 e IQR.
- Calcule os limites inferior e superior do IQR.
- Calcule média e desvio-padrão.
- Calcule o Z-Score do valor 30.
- Compare o que o IQR e o Z-Score indicam sobre o valor 30.
- Escreva uma conclusão curta explicando que técnicas diferentes podem analisar o mesmo dado por critérios diferentes.

```python
import numpy as np

dados = [10, 11, 12, 12, 13, 13, 14, 15, 30]

q1 = np.percentile(dados, 25)
q3 = np.percentile(dados, 75)
iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

media = np.mean(dados)
desvio = np.std(dados)

z_30 = (30 - media) / desvio

print(f"Q1: {q1:.2f}")
print(f"Q3: {q3:.2f}")
print(f"IQR: {iqr:.2f}")
print(f"Limite inferior: {limite_inferior:.2f}")
print(f"Limite superior: {limite_superior:.2f}")
print()
print(f"Média: {media:.2f}")
print(f"Desvio-padrão: {desvio:.2f}")
print(f"Z-Score de 30: {z_30:.2f}")
print()

if 30 > limite_superior:
    print("Pelo IQR, o valor 30 é considerado um outlier (está acima do limite superior).")
else:
    print("Pelo IQR, o valor 30 não é considerado um outlier.")

if abs(z_30) > 3:
    print("Pelo Z-Score, o valor 30 merece investigação (|Z| > 3).")
else:
    print("Pelo Z-Score, o valor 30 não ultrapassa o limite de |Z| > 3, "
          "mesmo sendo apontado pelo IQR.")
```

**Conclusão:** o IQR e o Z-Score podem discordar sobre o mesmo dado porque usam critérios
diferentes. O IQR se baseia na dispersão dos quartis centrais e é pouco sensível a valores extremos,
por isso aponta o 30 como outlier com mais facilidade. Já o Z-Score se baseia na média e no
desvio-padrão, que são fortemente influenciados pelo próprio valor extremo — isso "infla" o
desvio-padrão e pode reduzir o Z-Score do valor, mascarando o quanto ele é incomum. Ou seja, técnicas
diferentes enxergam o mesmo dado sob perspectivas diferentes, e por isso é importante não depender de
uma única métrica para decidir o que é ou não um outlier.

## Exercício 10 — Mini análise de eventos de segurança

- Calcule a média de Tentativas_Login.
- Calcule o desvio-padrão.
- Calcule o Z-Score de cada evento.
- Identifique eventos com |Z| > 3.
- Mostre uma mensagem final explicando por que um evento incomum em segurança pode ser justamente o dado mais importante da análise.

```python
import pandas as pd

dados = {
    "Evento": ["A", "B", "C", "D", "E", "F", "G"],
    "Tentativas_Login": [3, 4, 2, 5, 3, 4, 40]
}

df = pd.DataFrame(dados)

media = df["Tentativas_Login"].mean()
desvio = df["Tentativas_Login"].std(ddof=0)

df["Z_Score"] = (df["Tentativas_Login"] - media) / desvio

print(f"Média: {media:.2f}")
print(f"Desvio-padrão: {desvio:.2f}")
print()
print("DataFrame completo:")
print(df)

eventos_investigar = df[df["Z_Score"].abs() > 3]
print()
print("Eventos com |Z| > 3:")
print(eventos_investigar)
```

**Mensagem final:** em segurança, um evento incomum — como o evento G, com um número de
tentativas de login muito acima do padrão — é justamente o tipo de dado que mais importa. Enquanto em
outros contextos um outlier pode ser apenas ruído, em segurança ele pode representar um ataque de
força bruta, uma tentativa de invasão ou um comportamento malicioso em andamento. Por isso, valores
com |Z| > 3 não devem ser descartados: eles devem ser investigados com prioridade, pois raridade
estatística aqui pode significar risco real.
